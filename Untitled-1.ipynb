In this notebook, we will securely load our API keys, make our first HTTP POST request to Groq, and parse the nested JSON response.

```python
#TEST ENVIRONMENTAL VARIABLES
import os
from dotenv import load_dotenv

# Load variables from the local .env file
load_dotenv()

api_key = os.getenv("GROQ_API_KEY")
model = os.getenv("GROQ_MODEL")

# Verify the key exists (prints True if loaded correctly)
print("API Key loaded successfully:", api_key is not None)
print("Active Model:", model)
```

Now that our environment variables are ready, we define our endpoint URL, headers (with authorization), and the JSON body containing our prompt.

```python
#MAKE REQUEST
import requests

url = "https://api.groq.com/openai/v1/chat/completions"

headers = {
    "Authorization": "Bearer " + api_key,
    "Content-Type": "application/json"
}

body = {
    "model": model,
    "messages": [
        {"role": "user", "content": "Explain APIs to a beginner."}
    ]
}

# Send the POST request
r = requests.post(url, headers=headers, json=body)

# Check the HTTP status code (200 means success!)
print("HTTP Status Code:", r.status_code)
```

The server sends back a complex JSON dictionary. We need to navigate through the nested keys (`choices`, `message`, `content`) to pull out the AI's answer and check our token consumption.

```python
# Convert the raw response into a Python dictionary
data = r.json()

# Navigate the nested JSON structure to get the text response
answer = data["choices"][0]["message"]["content"]
print("--- AI Answer ---")
print(answer)

# Extract token usage statistics
tokens = data["usage"]["total_tokens"]
print(f"\nTotal Tokens Used: {tokens}")
```
